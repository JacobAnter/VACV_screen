```python
"""
The purpose of this Jupyter notebook is to round the values in the top-k
metrics tables.
"""
```

```python
from decimal import Decimal, ROUND_HALF_UP

import pandas as pd
```

# Rounding the Table with Shuffled Labels

```python
# Load the table into a Pandas DataFrame
top_k_metrics_shuffled_labels_path = (
    "model_trained_on_shuffled_labels_performance_top_k_comparison.tsv"
)

top_k_metrics_shuffled_labels_df = pd.read_csv(
    top_k_metrics_shuffled_labels_path,
    sep="\t"
)
```

```python
# Apply rounding to four decimal places
top_k_metrics_shuffled_labels_rounded_df = (
    top_k_metrics_shuffled_labels_df.map(
        lambda val: float(
            Decimal(str(val))
            .quantize(Decimal("0.0001"), rounding=ROUND_HALF_UP)
        )
    )
)
```

```python
# Save the DataFrame to disk
top_k_metrics_shuffled_labels_rounded_df.to_csv(
    "model_trained_on_shuffled_labels_performance_top_k_comparison_rounded.tsv",
    sep="\t"
)
```
