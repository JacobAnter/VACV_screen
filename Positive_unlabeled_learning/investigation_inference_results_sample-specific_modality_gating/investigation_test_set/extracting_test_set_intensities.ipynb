```python
"""
The purpose of this Jupyter notebook is to extract the test set
intensities from the TSV files storing the intensities for the entire
screen.
"""
```

```python
import os

import pandas as pd
```

# Creating a Subdirectory

```python
data_subdir = "test_set_intensities"

if not os.path.exists(data_subdir):
    os.makedirs(data_subdir)
```

# Loading Data

```python
# Load the TSV files storing the unrefined as well as the refined
# intensities
unrefined_early_antiviral_ints_path = (
    "/Users/jacobanter/Documents/Code/VACV_screen/Positive_unlabeled_"
    "learning/investigation_inference_results_sample-specific_"
    "modality_gating/intensity_scores/unrefined_early_int_scores_"
    "antiviral.tsv"
)
unrefined_early_antiviral_ints_df = pd.read_csv(
    unrefined_early_antiviral_ints_path,
    sep="\t"
)

unrefined_early_proviral_ints_path = (
    "/Users/jacobanter/Documents/Code/VACV_screen/Positive_unlabeled_"
    "learning/investigation_inference_results_sample-specific_modality_"
    "gating/intensity_scores/unrefined_early_int_scores_proviral.tsv"
)
unrefined_early_proviral_ints_df = pd.read_csv(
    unrefined_early_proviral_ints_path,
    sep="\t"
)

unrefined_late_antiviral_ints_path = (
    "/Users/jacobanter/Documents/Code/VACV_screen/Positive_unlabeled_"
    "learning/investigation_inference_results_sample-specific_modality_"
    "gating/intensity_scores/unrefined_late_int_scores_antiviral.tsv"
)
unrefined_late_antiviral_ints_df = pd.read_csv(
    unrefined_late_antiviral_ints_path,
    sep="\t"
)

unrefined_late_proviral_ints_path = (
    "/Users/jacobanter/Documents/Code/VACV_screen/Positive_unlabeled_"
    "learning/investigation_inference_results_sample-specific_modality_"
    "gating/intensity_scores/unrefined_late_int_scores_proviral.tsv"
)
unrefined_late_proviral_ints_df = pd.read_csv(
    unrefined_late_proviral_ints_path,
    sep="\t"
)


refined_early_antiviral_ints_path = (
    "/Users/jacobanter/Documents/Code/VACV_screen/Positive_unlabeled_"
    "learning/investigation_inference_results_sample-specific_modality_"
    "gating/intensity_scores/refined_early_int_scores_antiviral.tsv"
)
refined_early_antiviral_ints_df = pd.read_csv(
    refined_early_antiviral_ints_path,
    sep="\t"
)

refined_early_proviral_ints_path = (
    "/Users/jacobanter/Documents/Code/VACV_screen/Positive_unlabeled_"
    "learning/investigation_inference_results_sample-specific_modality_"
    "gating/intensity_scores/refined_early_int_scores_proviral.tsv"
)
refined_early_proviral_ints_df = pd.read_csv(
    refined_early_proviral_ints_path,
    sep="\t"
)

refined_late_antiviral_ints_path = (
    "/Users/jacobanter/Documents/Code/VACV_screen/Positive_unlabeled_"
    "learning/investigation_inference_results_sample-specific_modality_"
    "gating/intensity_scores/refined_late_int_scores_antiviral.tsv"
)
refined_late_antiviral_ints_df = pd.read_csv(
    refined_late_antiviral_ints_path,
    sep="\t"
)

refined_late_proviral_ints_path = (
    "/Users/jacobanter/Documents/Code/VACV_screen/Positive_unlabeled_"
    "learning/investigation_inference_results_sample-specific_modality_"
    "gating/intensity_scores/refined_late_int_scores_proviral.tsv"
)
refined_late_proviral_ints_df = pd.read_csv(
    refined_late_proviral_ints_path,
    sep="\t"
)

# Load the test set TSV file
test_set_path = (
    "/Users/jacobanter/Documents/Code/VACV_screen/Positive_unlabeled_"
    "learning/Train_validation_test_split/scientifically_sound_host_"
    "factor_set/all_host_factors/all_hf_formatted_data/all_hf_test_set_"
    "formatted_with_probs.tsv"
)

test_set_df = pd.read_csv(
    test_set_path,
    sep="\t"
)
```

# Stitching the DataFrames together

```python
# Antiviral (positive) and proviral (negative) intensities belonging to
# the same channel (early/late) and refinement regime
# (unrefined/refined) are stitched together
unrefined_early_ints_df = pd.concat(
    [
        unrefined_early_antiviral_ints_df,
        unrefined_early_proviral_ints_df
    ],
    ignore_index=True
)

unrefined_late_ints_df = pd.concat(
    [
        unrefined_late_antiviral_ints_df,
        unrefined_late_proviral_ints_df
    ],
    ignore_index=True
)

refined_early_ints_df = pd.concat(
    [
        refined_early_antiviral_ints_df,
        refined_early_proviral_ints_df
    ],
    ignore_index=True
)

refined_late_ints_df = pd.concat(
    [
        refined_late_antiviral_ints_df,
        refined_late_proviral_ints_df
    ],
    ignore_index=True
)
```

# Extracting Test Set Genes

```python
print(
    "Length of DataFrames prior to filtering:\n"
    f"{len(unrefined_early_ints_df)}, {len(unrefined_late_ints_df)}, "
    f"{len(refined_early_ints_df)}, {len(refined_late_ints_df)}"
)
```

```python
# Extract test set genes
test_set_genes = test_set_df["gene"].unique()

# Filter the intensity DataFrames to retain only test set genes
unrefined_early_ints_df = unrefined_early_ints_df[
    unrefined_early_ints_df["gene"].isin(test_set_genes)
]

unrefined_late_ints_df = unrefined_late_ints_df[
    unrefined_late_ints_df["gene"].isin(test_set_genes)
]

refined_early_ints_df = refined_early_ints_df[
    refined_early_ints_df["gene"].isin(test_set_genes)
]

refined_late_ints_df = refined_late_ints_df[
    refined_late_ints_df["gene"].isin(test_set_genes)
]
```

```python
print(
    "Length of DataFrames after filtering:\n"
    f"{len(unrefined_early_ints_df)}, {len(unrefined_late_ints_df)}, "
    f"{len(refined_early_ints_df)}, {len(refined_late_ints_df)}"
)
```

# Renaming the "Probability" Column

```python
unrefined_early_ints_df.rename(
    columns={"probability": "intensity"},
    inplace=True
)

unrefined_late_ints_df.rename(
    columns={"probability": "intensity"},
    inplace=True
)

refined_early_ints_df.rename(
    columns={"probability": "intensity"},
    inplace=True
)

refined_late_ints_df.rename(
    columns={"probability": "intensity"},
    inplace=True
)
```

# Saving the DataFrames to Disk

```python
unrefined_early_ints_df.to_csv(
    os.path.join(
        data_subdir,
        "unrefined_early_test_set_intensities.tsv"
    ),
    sep="\t",
    index=False
)

unrefined_late_ints_df.to_csv(
    os.path.join(
        data_subdir,
        "unrefined_late_test_set_intensities.tsv"
    ),
    sep="\t",
    index=False
)

refined_early_ints_df.to_csv(
    os.path.join(
        data_subdir,
        "refined_early_test_set_intensities.tsv"
    ),
    sep="\t",
    index=False
)

refined_late_ints_df.to_csv(
    os.path.join(
        data_subdir,
        "refined_late_test_set_intensities.tsv"
    ),
    sep="\t",
    index=False
)
```
